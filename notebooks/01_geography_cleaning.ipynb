```python
csv = pd.read_csv(
    "../raw/State_Region_corrected.csv"
)

csv = csv.rename(columns={
    "State / Union territory (UT)": "state_name",
    "Area (km2)": "area_km2",
    "Region": "region",
    "National Share (%)": "national_share_pct"
})
```

```python
print(csv.columns.tolist())
```

```python
geo_names = set(geo["state_name"])
csv_names = set(csv["state_name"])

print("GeoJSON but not CSV:")
print(sorted(geo_names - csv_names))

print("\nCSV but not GeoJSON:")
print(sorted(csv_names - geo_names))
```

```python
dim_state = geo.merge(
    csv,
    on="state_name",
    how="left",
    validate="one_to_one"
)
```

```python
print(dim_state.shape)
```

```python
print(
    dim_state[
        dim_state["area_km2"].isna()
    ][
        ["state_id", "state_name"]
    ]
)
```

```python
dim_state_tabular = dim_state[
    [
        "state_id",
        "state_name",
        "shapeISO",
        "shapeID",
        "area_km2",
        "region",
        "national_share_pct"
    ]
].copy()
```

```python
dim_state_geo = dim_state[
    [
        "state_id",
        "state_name",
        "shapeISO",
        "shapeID",
        "area_km2",
        "region",
        "national_share_pct",
        "geometry"
    ]
].copy()
```

```python
import os

os.makedirs("../clean", exist_ok=True)

dim_state_tabular.to_csv(
    "../clean/dim_state.csv",
    index=False
)

dim_state_geo.to_file(
    "../clean/dim_state.geojson",
    driver="GeoJSON"
)
```
